1. Solve step by step a random 4×4 linear system (use the first 4 digits of your cpf for the seed), describing each step of the GaussElimin and GaussRetro algorithm. Check the answer by calculating the residue.

```python
import numpy as np

#os primeiros 4 dígitos do CPF como seed: 0140
np.random.seed(140)

# Criar matriz aleatoria 4x4
A = np.random.rand(4, 4)
# Criar un vetor aleatorio 4x1
B = np.random.rand(4, 1)

print(f'A:{A}')
print(f'B: {B}')
```

```python
# Función para eliminación gaussiana
def gauss(a, b):
    n = len(b)
    a = a.copy()  # preserva a entrada
    b = b.copy()

    for k in range(0, n - 1):  # linha pivo
        for i in range(k + 1, n):  # recorre las filas / columna a eliminar
            m = a[i][k] / a[k][k]  # operación entre líneas

            a[i, k + 1:n] = a[i, k + 1:n] - m * a[k, k + 1:n]  # actualiza la línea i (vectorización)
            a[i, k] = 0  # que es 0 por definición
            b[i] = b[i] - m * b[k]  # actualiza el elemento de la columna

    return np.hstack([a, b])  # devuelve la matriz aumentada

gauss(A,B)
```

```python
# Función retro-sustitución
def GaussRetro(a, b):
    a = a.copy()
    b = b.copy()
    n = len(b)
    x = np.zeros(n)   #cria vetor de tamanho n
    x[n-1] = (b[n-1] / a[n-1, n-1]).item()   #calcula a incógnita no último elemento
    for k in range(n-2, -1, -1):
        x[k] = ((b[k] - np.dot(a[k, k+1:n], x[k+1:n])) / a[k, k]).item()   #item extrae el valor como escalar y no como array


    return np.transpose([x])  # Devolve vetor de soluções em forma de vetor coluna
```

```python
# Chamar função para fazer eliminação da matriz aumentada
M1 = gauss(A, B)
M1
```

```python
# uma vez que temos a matriz escalonada, separar a matriz A das incógnitas, e B
A1 = M1[:, 0:4]
B1 = M1[:, 4]

# chamar a função de calcular as incógnitas por retrosubstituição
x1 = GaussRetro(A1, B1)
print(f'X= {x1}')

# Calcular residuo
residuo = A @ x1 - B
print(f'\n Residuos:\n {residuo}')
```

2. Solve a random system 500×500 using the GaussSolve algorithm (do not show the result). Check the correctness of the result by calculating the residue norm and compare the required CPU time. Repeat the procedure using numpy's solver.

```python
#criar matriz 500x500
import numpy as np
np.random.seed(123)
N=500
A=np.random.rand(N,N)
B=np.random.rand(N,1)

```

```python
def GaussSolve(a,b):
  n=len(a)
  a=a.copy() #preserva a entrada
  b=b.copy()
  x=np.zeros(n)
  for k in range(0,n-1): #linha pivot
    for i in range(k+1,n): #percorre linhas / coluna a ser eliminada
      m=a[i][k]/a[k][k]  #operaçaõ entre linhas

      a[i,k+1:n]=a[i,k+1:n]-m*a[k,k+1:n] #atualiza elem da linha i (vetorização)
      a[i,k]=0   #que é zero por definição
      b[i]=b[i]-m*b[k]   #atualiza elem da coluna
    x[n-1]=(b[n-1]/a[n-1,n-1]).item()
  for k in range(n-2,-1,-1):
    x[k]=((b[k]-np.dot(a[k,k+1:n],x[k+1:n]))/a[k,k]).item()

  return np.transpose([x])
```

```python
x=GaussSolve(A,B) #obtendo a solução e mostrando apenas o residuo

residuo = np.linalg.norm(A @ x - B, ord=np.inf)
residuo
```

```python
# mostrar tempo de resolução do algoritmo
%%timeit
GaussSolve(A, B)
```

3. The condition number of a matrix A, denoted by κ(A), measures the sensitivity of the solution of a linear system Ax=b to such small perturbations in the matrix A or in the vector b. It is defined as:
κ(A)=∥A∥⋅∥A−1∥,
where ∥A∥ is the norm of the matrix (usually the 2-norm). A high condition number indicates that the system is ill-conditioned, i.e., small changes in the data can cause large variations in the solution. An example of an ill-conditioned matrix is ​​the Hilbert matrix Hn of order n, defined by
Hij=1i+j+1,i,j=0,1,…,n.

```python
# função para criar a matriz de Hilbert
def hilb(n):
    return np.array([[1/(i+j+1) for i in range(n)] for j in range(n)])
    return np.array([[1/(i+j+1) for i in range(n)] for j in range(n)])
```

```python
H = hilb(30) #matriz Hilbert 30x30
B = np.ones([30,1]) #criar matriz B com elementos = 1

# usar algoritmo de eliminação gaussiana e mostrando o residuo
X = GaussSolve(H, B)
residuo_gauss = np.linalg.norm(H @ X - B, ord=np.inf)
print(f'Residuo: {residuo_gauss}')

```
